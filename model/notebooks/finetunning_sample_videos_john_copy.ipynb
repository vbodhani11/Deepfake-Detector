```python
#!/usr/bin/env python3
"""
train_xception.py — Fine-tune XceptionNet (via timm) on face crops (real/fake)
Dataset layout (ImageFolder style):
crops/
  train/real/*.jpg
  train/fake/*.jpg
  val/real/*.jpg
  val/fake/*.jpg

Usage:
Run the train_model function with appropriate arguments.
"""

import argparse, os, random, time, math
from pathlib import Path

import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import DataLoader
from torchvision import datasets, transforms

# Import torch_xla
# import torch_xla
# import torch_xla.core.xla_model as xm

try:
    import timm
except Exception as e:
    raise SystemExit("Please install timm: pip install timm==1.*") from e

from sklearn.metrics import accuracy_score, f1_score, roc_auc_score, average_precision_score, log_loss

IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD  = [0.229, 0.224, 0.225]
```

```python
def seed_everything(seed: int = 42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

def build_loaders(data_dir, img_size=299, batch_size=64, num_workers=8):
    train_tf = transforms.Compose([
        transforms.RandomResizedCrop(img_size, scale=(0.9, 1.0)),
        transforms.RandomApply([transforms.GaussianBlur(3)], p=0.2),
        transforms.ColorJitter(0.1,0.1,0.1,0.05),
        transforms.ToTensor(),
        transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
    ])
    val_tf = transforms.Compose([
        transforms.Resize(img_size),
        transforms.CenterCrop(img_size),
        transforms.ToTensor(),
        transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
    ])
    train_ds = datasets.ImageFolder(os.path.join(data_dir, "train"), transform=train_tf)
    val_ds   = datasets.ImageFolder(os.path.join(data_dir, "val"),   transform=val_tf)
    train_loader = DataLoader(train_ds, batch_size=batch_size, shuffle=True,
                              num_workers=num_workers, pin_memory=True)
    val_loader   = DataLoader(val_ds,   batch_size=batch_size, shuffle=False,
                              num_workers=num_workers, pin_memory=True)
    return train_loader, val_loader, train_ds.classes
```

```python
def evaluate_frame_level(model, loader, device, alpha=1.0, threshold=0.5):
    """
    Evaluate model and compute metrics including weighted precision.

    Args:
        model: The model to evaluate
        loader: DataLoader for evaluation
        device: Device to run evaluation on
        alpha: Weight factor for false positives in weighted precision.
               alpha = (ratio of real:fake in organic traffic) / (ratio of real:fake in dataset)
               For example, if organic is 1M:1 and dataset is 1:1, alpha = 1M
               Default 1.0 means no weighting (standard precision)
        threshold: Classification threshold for fake class (default 0.5)

    Returns:
        Dictionary of metrics including weighted precision (wP)
    """
    from sklearn.metrics import confusion_matrix

    model.eval()
    all_logits, all_targets = [], []
    with torch.no_grad():
        for x, y in loader:
            x = x.to(device)
            logits = model(x)
            all_logits.append(logits.cpu())
            all_targets.append(y)
    logits = torch.cat(all_logits, dim=0)
    targets = torch.cat(all_targets, dim=0).numpy()
    probs = torch.softmax(logits, dim=1).numpy()[:,1]
    preds = (probs >= threshold).astype(np.int64)

    # Calculate confusion matrix
    # Assuming: 0 = real, 1 = fake
    cm = confusion_matrix(targets, preds)
    if cm.shape == (2, 2):
        TN, FP, FN, TP = cm.ravel()
    else:
        # Handle edge cases where one class is missing
        if len(np.unique(targets)) == 1:
            if targets[0] == 0:  # All real
                TN, FP, FN, TP = len(targets), 0, 0, 0
            else:  # All fake
                TN, FP, FN, TP = 0, 0, 0, len(targets)
        else:
            TN, FP, FN, TP = 0, 0, 0, 0

    metrics = {}
    metrics["acc"] = float(accuracy_score(targets, preds))
    metrics["f1"]  = float(f1_score(targets, preds))

    # Calculate standard precision and recall
    if TP + FP > 0:
        metrics["precision"] = float(TP / (TP + FP))
    else:
        metrics["precision"] = 0.0

    if TP + FN > 0:
        metrics["recall"] = float(TP / (TP + FN))
    else:
        metrics["recall"] = 0.0

    # Calculate weighted precision: wP = TP / (TP + α * FP)
    if TP + alpha * FP > 0:
        metrics["wP"] = float(TP / (TP + alpha * FP))
    else:
        metrics["wP"] = 0.0

    # Additional metrics
    try:
        metrics["auc"] = float(roc_auc_score(targets, probs))
        metrics["ap_fake"] = float(average_precision_score(targets, probs))
        # Calculate log loss (binary cross-entropy)
        # Clip probabilities to avoid log(0) issues
        probs_clipped = np.clip(probs, 1e-15, 1 - 1e-15)
        metrics["log_loss"] = float(log_loss(targets, probs_clipped))
    except Exception:
        metrics["auc"] = float("nan")
        metrics["ap_fake"] = float("nan")
        metrics["log_loss"] = float("nan")

    # Store confusion matrix components for debugging
    metrics["TP"] = int(TP)
    metrics["FP"] = int(FP)
    metrics["FN"] = int(FN)
    metrics["TN"] = int(TN)

    return metrics
```

```python
def save_checkpoint(state, is_best, out_dir, name="xception"):
    out_dir = Path(out_dir)
    out_dir.mkdir(parents=True, exist_ok=True)
    ckpt = out_dir / f"{name}_last_dec2.pt"
    torch.save(state, ckpt)
    if is_best:
        best = out_dir / f"{name}_best_dec2.pt"
        torch.save(state, best)
```

```python
def train_model(data_dir, out_dir, epochs_head=3, epochs_full=15, batch_size=64, img_size=299, num_workers=8, seed=42, lr_head=3e-4, lr_full=1e-4, weight_decay=1e-4, resume=True, alpha=1.0, use_class_weights=True):
    """
    Train Xception model with checkpoint resuming and weighted precision for model selection.

    Args:
        alpha: Weight factor for false positives in weighted precision.
               alpha = (ratio of real:fake in organic traffic) / (ratio of real:fake in dataset)
               For example, if organic traffic has 1M real:1 fake and dataset has 1:1, alpha = 1,000,000
               Default 1.0 means standard precision (no weighting).
               Higher alpha penalizes false positives more, which is important for class imbalance.
        use_class_weights: If True, use class weights in loss to penalize false positives more.
                          Weights are calculated automatically from dataset imbalance.
    """
    seed_everything(seed)

    # Use CUDA device if available, otherwise CPU
    device = "cuda" if torch.cuda.is_available() else "cpu"
    print(f"Using device: {device}")
    print(f"Using weighted precision with alpha={alpha} for model selection")
    print(f"  (alpha = ratio of real:fake in organic traffic / ratio of real:fake in dataset)")
    print(f"  (higher alpha penalizes false positives more, important for class imbalance)")

    train_loader, val_loader, classes = build_loaders(data_dir, img_size, batch_size, num_workers)
    print(f"Classes: {classes} (expect ['fake', 'real'] or similar two classes)")

    # Calculate class weights if requested (to penalize false positives more)
    class_weights = None
    if use_class_weights:
        # Count samples per class in training set
        from collections import Counter
        all_labels = []
        for _, labels in train_loader:
            all_labels.extend(labels.numpy())
        label_counts = Counter(all_labels)

        # Calculate weights: inverse frequency, normalized
        # Higher weight for minority class (real) to reduce false positives
        total_samples = sum(label_counts.values())
        num_classes = len(label_counts)
        class_weights = torch.FloatTensor([
            total_samples / (num_classes * label_counts[i]) if i in label_counts else 1.0
            for i in range(num_classes)
        ]).to(device)

        print(f"Class weights: {class_weights.cpu().numpy()} (higher weight = more penalty for misclassification)")
        print(f"  This will penalize false positives (real->fake) more during training")
    else:
        print("Not using class weights (standard CrossEntropyLoss)")

    # Check for existing checkpoint
    out_dir_path = Path(out_dir)
    checkpoint_path = out_dir_path / "xception_last.pt"
    start_epoch = 0
    best_val = -1.0
    resume_from_phase2 = False
    checkpoint = None

    if resume and checkpoint_path.exists():
        print(f"Found checkpoint at {checkpoint_path}. Loading...")
        checkpoint = torch.load(checkpoint_path, map_location=device)

        # Build model
        model = timm.create_model("xception", pretrained=False, num_classes=2)
        model.load_state_dict(checkpoint['model_state'])
        model = model.to(device)

        # Restore training state
        start_epoch = checkpoint.get('epoch', 0)
        # Restore best weighted precision, fallback to accuracy for old checkpoints
        val_metrics_ckpt = checkpoint.get('val_metrics', {})
        best_val = val_metrics_ckpt.get('wP', val_metrics_ckpt.get('acc', -1.0))

        # Check which phase we were in
        if start_epoch > epochs_head:
            # We were in Phase 2
            resume_from_phase2 = True
            print(f"Resuming from Phase 2, epoch {start_epoch - epochs_head}/{epochs_full} (total epoch {start_epoch})")
        else:
            # We were in Phase 1
            print(f"Resuming from Phase 1, epoch {start_epoch}/{epochs_head}")

        # Restore args if they exist (for validation)
        saved_args = checkpoint.get('args', {})
        if saved_args:
            print(f"Checkpoint was saved with args: epochs_head={saved_args.get('epochs_head')}, epochs_full={saved_args.get('epochs_full')}")
    else:
        # Build model from scratch
        print("No checkpoint found. Starting training from scratch with pretrained ImageNet weights.")
        model = timm.create_model("xception", pretrained=True, num_classes=2)
        model = model.to(device)

    # Phase 1: freeze backbone, train head
    head_params = []
    for n, p in model.named_parameters():
        if "fc" in n or "classifier" in n:
            head_params.append(p)
            p.requires_grad = True
        else:
            p.requires_grad = False

    opt = torch.optim.AdamW(head_params, lr=lr_head, weight_decay=weight_decay)
    crit = nn.CrossEntropyLoss(weight=class_weights) if class_weights is not None else nn.CrossEntropyLoss()

    # Load optimizer state if resuming from Phase 1
    if resume and checkpoint is not None and not resume_from_phase2:
        if 'optimizer_state' in checkpoint:
            opt.load_state_dict(checkpoint['optimizer_state'])
            print("Loaded optimizer state from checkpoint.")

    def train_one_epoch(epoch, model, loader, optimizer, criterion, device):
        model.train()
        losses = []
        for x, y in loader:
            x, y = x.to(device), y.to(device)
            optimizer.zero_grad(set_to_none=True)
            logits = model(x)
            loss = criterion(logits, y)
            loss.backward()
            optimizer.step()
            losses.append(loss.item())
        return float(np.mean(losses))

    # Phase 1: freeze backbone, train head (skip if resuming from Phase 2)
    if not resume_from_phase2:
        print("==> Phase 1: training the head")
        phase1_start = start_epoch + 1 if start_epoch < epochs_head else epochs_head + 1
        for epoch in range(phase1_start, epochs_head+1):
            tr_loss = train_one_epoch(epoch, model, train_loader, opt, crit, device)
            val_metrics = evaluate_frame_level(model, val_loader, device, alpha=alpha)
            # Use weighted precision for model selection (better for class imbalance)
            score = val_metrics["wP"]
            is_best = score > best_val
            if is_best: best_val = score
            save_checkpoint({
                "epoch": epoch,
                "model_state": model.state_dict(),
                "optimizer_state": opt.state_dict(),
                "val_metrics": val_metrics,
                "args": {
                    "data_dir": data_dir, "out_dir": out_dir, "epochs_head": epochs_head,
                    "epochs_full": epochs_full, "batch_size": batch_size, "img_size": img_size,
                    "num_workers": num_workers, "seed": seed, "lr_head": lr_head,
                    "lr_full": lr_full, "weight_decay": weight_decay, "alpha": alpha
                },
            }, is_best, out_dir, name="xception")
            print(f"[Head {epoch}/{epochs_head}] loss={tr_loss:.4f}  wP={score:.4f}  acc={val_metrics['acc']:.4f}  prec={val_metrics.get('precision', 0):.4f}  rec={val_metrics.get('recall', 0):.4f}  log_loss={val_metrics.get('log_loss', float('nan')):.4f}  TP={val_metrics.get('TP', 0)} FP={val_metrics.get('FP', 0)}")
    else:
        print("==> Phase 1: already completed, skipping to Phase 2")

    # Phase 2: unfreeze all, train
    for p in model.parameters():
        p.requires_grad = True
    opt = torch.optim.AdamW(model.parameters(), lr=lr_full, weight_decay=weight_decay)

    # Load optimizer state if resuming from Phase 2
    if resume and checkpoint is not None and resume_from_phase2:
        if 'optimizer_state' in checkpoint:
            opt.load_state_dict(checkpoint['optimizer_state'])
            print("Loaded optimizer state from checkpoint.")

    print("==> Phase 2: fine-tuning all layers")
    total_epochs = epochs_full

    # Calculate starting epoch for Phase 2
    if resume_from_phase2:
        phase2_start = (start_epoch - epochs_head) + 1
    else:
        phase2_start = 1

    for epoch in range(phase2_start, total_epochs+1):
        tr_loss = train_one_epoch(epoch, model, train_loader, opt, crit, device)
        val_metrics = evaluate_frame_level(model, val_loader, device, alpha=alpha)
        # Use weighted precision for model selection (better for class imbalance)
        score = val_metrics["wP"]
        is_best = score > best_val
        if is_best: best_val = score
        save_checkpoint({
            "epoch": epochs_head + epoch,
            "model_state": model.state_dict(),
            "optimizer_state": opt.state_dict(),
            "val_metrics": val_metrics,
            "args": {
                "data_dir": data_dir, "out_dir": out_dir, "epochs_head": epochs_head,
                "epochs_full": epochs_full, "batch_size": batch_size, "img_size": img_size,
                "num_workers": num_workers, "seed": seed, "lr_head": lr_head,
                "lr_full": lr_full, "weight_decay": weight_decay, "alpha": alpha
            },
        }, is_best, out_dir, name="xception")
        print(f"[FT {epoch}/{total_epochs}] loss={tr_loss:.4f}  wP={score:.4f}  acc={val_metrics['acc']:.4f}  prec={val_metrics.get('precision', 0):.4f}  rec={val_metrics.get('recall', 0):.4f}  log_loss={val_metrics.get('log_loss', float('nan')):.4f}  TP={val_metrics.get('TP', 0)} FP={val_metrics.get('FP', 0)}")

    print(f"Training complete. Best weighted precision (wP): {best_val:.4f}")

# Example usage (you can call this function in a separate cell):
data_dir = '/content/drive/MyDrive/DeepfakeData/SampleData/crops' # Replace with your crops directory
out_dir = '/content/drive/MyDrive/DeepfakeData/SampleData/runs/xception_dfdc' # Replace with your desired output directory for runs

# Calculate alpha from your dataset if needed:
# For example, if your dataset has 1000 real:1000 fake (1:1 ratio) and organic traffic is 1M:1,
# then alpha = (1,000,000/1) / (1000/1000) = 1,000,000
# If unsure, start with alpha=1.0 (standard precision) or alpha=100-1000 for moderate imbalance
```

```python
# ============================================================================
# VIDEO-LEVEL PREDICTION: Combine Frames by Video Name
# ============================================================================
# Groups frames by video name and aggregates predictions to determine if video is fake or real

def predict_video_from_frames(video_name, frames_dir='/content/drive/MyDrive/DeepfakeData/CompleteData/organized/crops/test', threshold=0.85):
    """
    Predict if a video is fake or real by combining all frames from that video.

    Args:
        video_name: Name of the video (e.g., 'ahqqqilsxt')
        frames_dir: Base directory containing frames (looks in both real_videos and fake_videos)
        threshold: Probability threshold for frame classification

    Returns:
        Dictionary with video-level prediction and frame statistics
    """
    frames_base = Path(frames_dir)

    # Search for frames in both real and fake folders
    real_pattern = frames_base / 'real_videos' / f'{video_name}_*.jpg'
    fake_pattern = frames_base / 'fake_videos' / f'{video_name}_*.jpg'

    real_frames = glob.glob(str(real_pattern))
    fake_frames = glob.glob(str(fake_pattern))

    # Determine ground truth and get frame files
    if real_frames and not fake_frames:
        frame_files = real_frames
        ground_truth = 'real'
    elif fake_frames and not real_frames:
        frame_files = fake_frames
        ground_truth = 'fake'
    elif real_frames and fake_frames:
        frame_files = real_frames if len(real_frames) >= len(fake_frames) else fake_frames
        ground_truth = 'real' if len(real_frames) >= len(fake_frames) else 'fake'
    else:
        return {
            'error': f'No frames found for video "{video_name}"',
            'video_name': video_name,
            'prediction': None
        }

    frame_files.sort()

    # Load and preprocess all frames
    frames = []
    for frame_path in frame_files:
        img = Image.open(frame_path).convert('RGB')
        frames.append(val_tf(img))

    frames_tensor = torch.stack(frames).to(device)

    # Get frame-level predictions
    model.eval()
    with torch.no_grad():
        logits = model(frames_tensor)
        probs = torch.softmax(logits, dim=1)[:, 0].cpu().numpy()  # Class 0 = FAKE

    # Frame-level statistics
    total_frames = len(probs)
    avg_fake_prob = float(np.mean(probs))
    max_fake_prob = float(np.max(probs))
    min_fake_prob = float(np.min(probs))

    # Count frames above threshold
    frames_above_threshold = np.sum(probs >= threshold)
    fake_ratio = frames_above_threshold / total_frames

    # Video-level prediction: Majority vote
    # If >50% of frames are predicted as fake, video is fake
    video_prediction = 'fake' if fake_ratio > 0.5 else 'real'

    # Check if correct
    is_correct = (video_prediction == ground_truth)

    return {
        'video_name': video_name,
        'ground_truth': ground_truth,
        'prediction': video_prediction,
        'is_correct': is_correct,
        'total_frames': total_frames,
        'avg_fake_prob': avg_fake_prob,
        'max_fake_prob': max_fake_prob,
        'min_fake_prob': min_fake_prob,
        'frames_above_threshold': int(frames_above_threshold),
        'fake_ratio': fake_ratio,
        'frame_probabilities': probs.tolist()
    }

def analyze_all_videos_in_directory(frames_dir='/content/drive/MyDrive/DeepfakeData/CompleteData/organized/crops/test', threshold=0.85):
    """
    Analyze all videos in a directory by combining frames by video name.

    Args:
        frames_dir: Directory containing frames (with real_videos and fake_videos subdirs)
        threshold: Probability threshold for frame classification

    Returns:
        Dictionary with results for all videos
    """
    frames_base = Path(frames_dir)

    # Get all unique video names from both folders
    all_video_names = set()

    # From fake_videos folder
    fake_dir = frames_base / 'fake_videos'
    if fake_dir.exists():
        for frame_file in fake_dir.glob('*.jpg'):
            video_name = frame_file.stem.rsplit('_', 1)[0]
            all_video_names.add(video_name)

    # From real_videos folder
    real_dir = frames_base / 'real_videos'
    if real_dir.exists():
        for frame_file in real_dir.glob('*.jpg'):
            video_name = frame_file.stem.rsplit('_', 1)[0]
            all_video_names.add(video_name)

    all_video_names = sorted(list(all_video_names))

    print(f"Found {len(all_video_names)} unique videos")
    print(f"Analyzing videos with threshold={threshold}...")
    print("="*80)

    # Analyze each video
    results = []
    for video_name in all_video_names:
        result = predict_video_from_frames(video_name, frames_dir, threshold)
        if 'error' not in result:
            results.append(result)

    return results

# ============================================================================
# Analyze Test Set Videos
# ============================================================================
print("="*80)
print("VIDEO-LEVEL ANALYSIS: Combining Frames by Video Name")
print("="*80)
print("This will group all frames by video name and determine if each video is fake or real")
print("="*80)

# Analyze all videos in test set
test_results = analyze_all_videos_in_directory(
    frames_dir='/content/drive/MyDrive/DeepfakeData/CompleteData/organized/crops/test',
    threshold=0.85
)

# Separate by ground truth
fake_videos = [r for r in test_results if r['ground_truth'] == 'fake']
real_videos = [r for r in test_results if r['ground_truth'] == 'real']

# Calculate metrics
fake_correct = sum(1 for r in fake_videos if r['is_correct'])
fake_total = len(fake_videos)
fake_accuracy = fake_correct / fake_total if fake_total > 0 else 0

real_correct = sum(1 for r in real_videos if r['is_correct'])
real_total = len(real_videos)
real_accuracy = real_correct / real_total if real_total > 0 else 0

total_correct = fake_correct + real_correct
total_videos = len(test_results)
overall_accuracy = total_correct / total_videos if total_videos > 0 else 0

# Print summary
print(f"\n{'='*80}")
print("VIDEO-LEVEL RESULTS (Frames Combined by Video Name)")
print(f"{'='*80}")
print(f"\nTotal Videos Analyzed: {total_videos}")
print(f"  Fake videos: {fake_total}")
print(f"  Real videos: {real_total}")

print(f"\nFAKE Videos:")
print(f"  Correctly identified as FAKE: {fake_correct}/{fake_total}")
print(f"  Accuracy: {fake_accuracy:.4f} ({fake_accuracy*100:.2f}%)")

print(f"\nREAL Videos:")
print(f"  Correctly identified as REAL: {real_correct}/{real_total}")
print(f"  Accuracy: {real_accuracy:.4f} ({real_accuracy*100:.2f}%)")

print(f"\nOVERALL:")
print(f"  Correct predictions: {total_correct}/{total_videos}")
print(f"  Overall accuracy: {overall_accuracy:.4f} ({overall_accuracy*100:.2f}%)")

# Show detailed results for each video
print(f"\n{'='*80}")
print("DETAILED RESULTS FOR EACH VIDEO:")
print(f"{'='*80}")
print(f"{'Video Name':<20} {'GT':<6} {'Pred':<6} {'Correct':<8} {'Avg Prob':<10} {'Fake Ratio':<12} {'Frames':<8}")
print("-"*80)

for result in sorted(test_results, key=lambda x: (x['ground_truth'], x['video_name'])):
    correct_str = "✓ YES" if result['is_correct'] else "✗ NO"
    print(f"{result['video_name']:<20} {result['ground_truth']:<6} {result['prediction']:<6} {correct_str:<8} {result['avg_fake_prob']:<10.4f} {result['fake_ratio']:<12.4f} {result['total_frames']:<8}")

# Show failed videos
failed_videos = [r for r in test_results if not r['is_correct']]
if failed_videos:
    print(f"\n{'='*80}")
    print(f"FAILED VIDEOS ({len(failed_videos)}):")
    print(f"{'='*80}")
    for r in failed_videos:
        print(f"  {r['video_name']}: Ground truth={r['ground_truth']}, Predicted={r['prediction']}")
        print(f"    Avg fake prob: {r['avg_fake_prob']:.4f}, Fake ratio: {r['fake_ratio']:.4f}, Frames: {r['total_frames']}")

print("\n" + "="*80)
print("Analysis complete!")
print("="*80)
```

```python
# ============================================================================
# DFDC WINNER'S CONFIDENT STRATEGY
# ============================================================================
# Implementation of the aggregation strategy from DFDC challenge winner
# This uses high-confidence frames instead of simple averaging

import numpy as np

def confident_strategy(pred, t=0.8):
    """
    DFDC winner's confident aggregation strategy.

    Args:
        pred: Array of frame-level fake probabilities
        t: Threshold for high-confidence fake predictions (default 0.8)

    Returns:
        Aggregated fake probability for the video
    """
    pred = np.array(pred)
    sz = len(pred)
    fakes = np.count_nonzero(pred > t)

    # If many frames are detected as fakes with high probability
    if fakes > sz // 2.5 and fakes > 11:
        return np.mean(pred[pred > t])
    # If most frames are clearly real (low fake probability)
    elif np.count_nonzero(pred < 0.2) > 0.9 * sz:
        return np.mean(pred[pred < 0.2])
    # Otherwise, use mean of all frames
    else:
        return np.mean(pred)

def predict_video_with_confident_strategy(video_name, frames_dir='/content/drive/MyDrive/DeepfakeData/CompleteData/organized/crops/test', threshold=0.85, confident_t=0.8):
    """
    Predict video using DFDC winner's confident strategy.

    Args:
        video_name: Name of the video
        frames_dir: Directory containing frames
        threshold: Final classification threshold (for fake/real decision)
        confident_t: Threshold for high-confidence frames in aggregation (default 0.8)

    Returns:
        Dictionary with prediction and statistics
    """
    frames_base = Path(frames_dir)

    # Search for frames
    real_pattern = frames_base / 'real_videos' / f'{video_name}_*.jpg'
    fake_pattern = frames_base / 'fake_videos' / f'{video_name}_*.jpg'

    real_frames = glob.glob(str(real_pattern))
    fake_frames = glob.glob(str(fake_pattern))

    # Determine ground truth and get frame files
    if real_frames and not fake_frames:
        frame_files = real_frames
        ground_truth = 'real'
    elif fake_frames and not real_frames:
        frame_files = fake_frames
        ground_truth = 'fake'
    elif real_frames and fake_frames:
        frame_files = real_frames if len(real_frames) >= len(fake_frames) else fake_frames
        ground_truth = 'real' if len(real_frames) >= len(fake_frames) else 'fake'
    else:
        return {
            'error': f'No frames found for video "{video_name}"',
            'video_name': video_name,
            'prediction': None
        }

    frame_files.sort()

    # Load and preprocess all frames
    frames = []
    for frame_path in frame_files:
        img = Image.open(frame_path).convert('RGB')
        frames.append(val_tf(img))

    frames_tensor = torch.stack(frames).to(device)

    # Get frame-level predictions
    model.eval()
    with torch.no_grad():
        logits = model(frames_tensor)
        probs = torch.softmax(logits, dim=1)[:, 0].cpu().numpy()  # Class 0 = FAKE

    # Apply confident strategy
    aggregated_prob = confident_strategy(probs, t=confident_t)

    # Video-level prediction
    video_prediction = 'fake' if aggregated_prob >= threshold else 'real'

    # Statistics
    total_frames = len(probs)
    avg_fake_prob = float(np.mean(probs))
    max_fake_prob = float(np.max(probs))
    min_fake_prob = float(np.min(probs))
    frames_above_threshold = np.sum(probs >= threshold)
    fake_ratio = frames_above_threshold / total_frames

    # Check which strategy was used
    sz = len(probs)
    fakes = np.count_nonzero(probs > confident_t)
    low_conf = np.count_nonzero(probs < 0.2)

    if fakes > sz // 2.5 and fakes > 11:
        strategy_used = f"High-confidence (>{confident_t}): {fakes} frames"
    elif low_conf > 0.9 * sz:
        strategy_used = f"Low-confidence (<0.2): {low_conf} frames"
    else:
        strategy_used = "Simple mean (all frames)"

    is_correct = (video_prediction == ground_truth)

    return {
        'video_name': video_name,
        'ground_truth': ground_truth,
        'prediction': video_prediction,
        'is_correct': is_correct,
        'total_frames': total_frames,
        'aggregated_prob': float(aggregated_prob),  # From confident strategy
        'avg_fake_prob': avg_fake_prob,  # Simple mean (for comparison)
        'max_fake_prob': max_fake_prob,
        'min_fake_prob': min_fake_prob,
        'frames_above_threshold': int(frames_above_threshold),
        'fake_ratio': fake_ratio,
        'strategy_used': strategy_used,
        'frame_probabilities': probs.tolist()
    }

# ============================================================================
# Test Confident Strategy on All Videos
# ============================================================================
print("="*80)
print("TESTING DFDC WINNER'S CONFIDENT STRATEGY")
print("="*80)
print("This uses high-confidence frames instead of simple averaging")
print("="*80)

# Get all video names
frames_base = Path('/content/drive/MyDrive/DeepfakeData/CompleteData/organized/crops/test')
all_video_names = set()

fake_dir = frames_base / 'fake_videos'
if fake_dir.exists():
    for frame_file in fake_dir.glob('*.jpg'):
        video_name = frame_file.stem.rsplit('_', 1)[0]
        all_video_names.add(video_name)

real_dir = frames_base / 'real_videos'
if real_dir.exists():
    for frame_file in real_dir.glob('*.jpg'):
        video_name = frame_file.stem.rsplit('_', 1)[0]
        all_video_names.add(video_name)

all_video_names = sorted(list(all_video_names))

print(f"Testing {len(all_video_names)} videos with confident strategy...")
print("="*80)

# Test with confident strategy
confident_results = []
for video_name in all_video_names:
    result = predict_video_with_confident_strategy(video_name, threshold=0.85, confident_t=0.8)
    if 'error' not in result:
        confident_results.append(result)

# Separate by ground truth
fake_videos_conf = [r for r in confident_results if r['ground_truth'] == 'fake']
real_videos_conf = [r for r in confident_results if r['ground_truth'] == 'real']

# Calculate metrics
fake_correct_conf = sum(1 for r in fake_videos_conf if r['is_correct'])
fake_total_conf = len(fake_videos_conf)
fake_accuracy_conf = fake_correct_conf / fake_total_conf if fake_total_conf > 0 else 0

real_correct_conf = sum(1 for r in real_videos_conf if r['is_correct'])
real_total_conf = len(real_videos_conf)
real_accuracy_conf = real_correct_conf / real_total_conf if real_total_conf > 0 else 0

total_correct_conf = fake_correct_conf + real_correct_conf
total_videos_conf = len(confident_results)
overall_accuracy_conf = total_correct_conf / total_videos_conf if total_videos_conf > 0 else 0

# Print results
print(f"\n{'='*80}")
print("RESULTS WITH CONFIDENT STRATEGY (threshold=0.85, confident_t=0.8)")
print(f"{'='*80}")
print(f"\nTotal Videos: {total_videos_conf}")
print(f"  Fake videos: {fake_total_conf}")
print(f"  Real videos: {real_total_conf}")

print(f"\nFAKE Videos:")
print(f"  Correctly identified as FAKE: {fake_correct_conf}/{fake_total_conf}")
print(f"  Accuracy: {fake_accuracy_conf:.4f} ({fake_accuracy_conf*100:.2f}%)")

print(f"\nREAL Videos:")
print(f"  Correctly identified as REAL: {real_correct_conf}/{real_total_conf}")
print(f"  Accuracy: {real_accuracy_conf:.4f} ({real_accuracy_conf*100:.2f}%)")

print(f"\nOVERALL:")
print(f"  Correct predictions: {total_correct_conf}/{total_videos_conf}")
print(f"  Overall accuracy: {overall_accuracy_conf:.4f} ({overall_accuracy_conf*100:.2f}%)")

# Compare with previous results (simple majority vote)
print(f"\n{'='*80}")
print("COMPARISON: Confident Strategy vs Simple Majority Vote")
print(f"{'='*80}")
print(f"{'Method':<30} {'Fake Acc':<12} {'Real Acc':<12} {'Overall Acc':<12}")
print("-"*70)
print(f"{'Simple Majority Vote':<30} {0.9697:<12.4f} {0.8750:<12.4f} {0.9512:<12.4f}")
print(f"{'Confident Strategy':<30} {fake_accuracy_conf:<12.4f} {real_accuracy_conf:<12.4f} {overall_accuracy_conf:<12.4f}")

improvement = overall_accuracy_conf - 0.9512
if improvement > 0:
    print(f"\n✓ Improvement: +{improvement:.4f} ({improvement*100:.2f}%)")
elif improvement < 0:
    print(f"\n✗ Decrease: {improvement:.4f} ({improvement*100:.2f}%)")
else:
    print(f"\n= No change")

# Show detailed results for failed videos (if any)
failed_videos_conf = [r for r in confident_results if not r['is_correct']]
if failed_videos_conf:
    print(f"\n{'='*80}")
    print(f"FAILED VIDEOS WITH CONFIDENT STRATEGY ({len(failed_videos_conf)}):")
    print(f"{'='*80}")
    for r in failed_videos_conf:
        print(f"  {r['video_name']}: GT={r['ground_truth']}, Pred={r['prediction']}")
        print(f"    Aggregated prob: {r['aggregated_prob']:.4f} (vs simple mean: {r['avg_fake_prob']:.4f})")
        print(f"    Strategy: {r['strategy_used']}")

# Show which strategy was used for each video
print(f"\n{'='*80}")
print("STRATEGY USAGE BREAKDOWN:")
print(f"{'='*80}")
strategy_counts = {}
for r in confident_results:
    strategy = r['strategy_used']
    if strategy not in strategy_counts:
        strategy_counts[strategy] = 0
    strategy_counts[strategy] += 1

for strategy, count in sorted(strategy_counts.items()):
    print(f"  {strategy}: {count} videos")

print("\n" + "="*80)
print("Analysis complete!")
print("="*80)
```
